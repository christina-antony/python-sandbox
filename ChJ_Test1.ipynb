```python
#sum of odd and even position numbers
num=input("enter positive number ")
odd=int(0)
even=int(0)
for n in num:
    if int(n)%2 == 0:
        odd=odd+int(n)
    else:
        even=even+int(n)
print("sum of odd numbers: ",odd,"sum of even numbers: ",even)
```

```python
def calculate_discount(order_amount: float):
    """Determines the discount percentage based on the order amount."""
    order_amount=float(order_amount)
    print(type(order_amount))
    if order_amount >= 100:
        print("20% discount applied")
        amt=order_amount-order_amount*0.20
        return amt
    elif order_amount >= 50:
        print("10% discount applied")
        amt=order_amount-order_amount*0.10
        return amt
    else:
        print("no discount applied")
        amt=order_amount
        return amt
```
